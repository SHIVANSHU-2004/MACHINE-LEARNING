```python

# Functions and File Handling
# 1. Function to Add Two Numbers
def add_numbers(a, b):
    return a + b
result = add_numbers(5, 3)
print("Sum:", result)

# 2. Function with Default Arguments
def greet(name="Guest"):
    print("Hello,", name)
greet("Alice")
greet()
# 3. Function with Positional and Keyword Arguments
def rectangle_area(length, width):
    return length * width
print("Area (Positional):", rectangle_area(5, 3))
print("Area (Keyword):", rectangle_area(width=4, length=6))
# 4. Lambda Function
multiply = lambda x, y: x * y
print("Multiplication:", multiply(4, 3))
# 5. Writing to a File
file = open("sample.txt", "w")
file.write("Hello, this is a sample file.\n")
file.write("Python file handling is simple!")
file.close()
print("\nData written to file successfully.")
# 6. Reading from a File
file = open("sample.txt", "r")
content = file.read()
print("\nFile Content:")
print(content)
file.close()
# 7. Appending Data to a File
file = open("sample.txt", "a")
file.write("\nAppending new content to the file.")
file.close()
# Reading Updated Content
file = open("sample.txt", "r")
print("\nUpdated File Content:")
print(file.read())
file.close()
# 8. Reading File Line by Line
print("\nReading File Line by Line:")
file = open("sample.txt", "r")
for line in file:
    print(line.strip())

file.close()
```
