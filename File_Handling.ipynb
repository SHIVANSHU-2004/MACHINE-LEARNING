```python
# ==================================================
# DICTIONARIES AND SETS IN PYTHON
# ==================================================

# -------------------------------
# 1. DICTIONARIES IN PYTHON
# -------------------------------

# Creating a dictionary
student_scores = {
    "Alice": 85,
    "Bob": 78,
    "Charlie": 92,
    "David": 88
}

print("Student Scores Dictionary:", student_scores)

# Accessing values
print("Alice's Score:", student_scores["Alice"])
print("Charlie's Score:", student_scores.get("Charlie"))
print("Eve's Score:", student_scores.get("Eve", "Not Found"))

# Adding and Updating Elements
student_scores["Eve"] = 95
print("\nAfter Adding Eve:", student_scores)

student_scores["Bob"] = 80
print("After Updating Bob's Score:", student_scores)

# Removing Elements
del student_scores["Alice"]
print("After Deleting Alice:", student_scores)

removed_score = student_scores.pop("Charlie")
print("Removed Charlie's Score:", removed_score)
print("After Removing Charlie:", student_scores)

# Iterating Through Dictionary
print("\nDictionary Keys:")
for student in student_scores:
    print(student)

print("\nDictionary Values:")
for score in student_scores.values():
    print(score)

print("\nDictionary Key-Value Pairs:")
for student, score in student_scores.items():
    print(student, ":", score)

# Dictionary Methods
print("\nIs Bob in Dictionary?", "Bob" in student_scores)
print("All Students:", list(student_scores.keys()))
print("All Scores:", list(student_scores.values()))
print("All Items:", list(student_scores.items()))

# -------------------------------
# 2. SETS IN PYTHON
# -------------------------------

# Creating a Set
numbers = {1, 2, 3, 4, 5, 5, 2}
print("\nNumbers Set:", numbers)

# Adding and Removing Elements
numbers.add(6)
print("After Adding 6:", numbers)

numbers.remove(3)
print("After Removing 3:", numbers)

numbers.discard(10)
print("After Discarding 10:", numbers)

removed_element = numbers.pop()
print("Removed Element:", removed_element)
print("After Pop:", numbers)

# Set Operations
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("\nSet A:", A)
print("Set B:", B)

print("Union:", A | B)
print("Intersection:", A & B)
print("Difference (A - B):", A - B)
print("Symmetric Difference:", A ^ B)

# Membership Testing
print("\nIs 2 in Set A?", 2 in A)
print("Is 6 in Set A?", 6 in A)

# Converting List to Set
num_list = [1, 2, 3, 4, 4, 2, 5, 6, 6]

unique_numbers = set(num_list)
print("\nOriginal List:", num_list)
print("Unique Numbers:", unique_numbers)

# Clearing Dictionary at the End
student_scores.clear()
print("\nDictionary After Clearing:", student_scores)

print("\nProgram Executed Successfully!")
```
