```python
#Declaring variables
name = "Alice" # String
age = 25 # Integer
height = 5.6 # Float
is_student = True # Boolean
# Printing variable values
print("Name:", name)
print("Age:", age)
print("Height:", height)
print("Is student:", is_student)
print(name, "is of type", type(name))
# Dictionary (key-value pairs)
student = {"name": "Alice", "age": 25, "grade": "A"}
print(student, "is of type", type(student))
a = 10
b = 3
# Addition
print("Addition:", a + b)
# Subtraction
print("Subtraction:", a - b)
# Multiplication
print("Multiplication:", a * b)
# Division
print("Division:", a / b)
# comparision operators
print("a > b:", a > b)
print("a < b:", a < b)
# logical Opearator
p = True
q = False
print("p AND q:", p and q) # True if both are True
print("p OR q:", p or q) # True if at least one is True
# Converting an integer to a string
num = 100
str_num = str(num)
print(str_num, "is of type", type(str_num))
# Converting a float to an integer
float_val = 10.99
int_from_float = int(float_val)
print(int_from_float, "is of type", type(int_from_float))
# Converting a number to a Boolean
bool_val = bool(0) # False because 0 is considered False
print(bool_val)
```
