```python
# SCATTER PLOT VISUALIZATION FOR EDA
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# 1. Load Dataset
df = pd.read_csv("/content/sample_data/california_housing_train.csv")

print("First Five Rows:")
print(df.head())

print("\nSummary Statistics:")
print(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())


plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='median_income',
    y='median_house_value',
    color='blue'
)

plt.title('Median House Value vs Median Income')
plt.xlabel('Median Income')
plt.ylabel('Median House Value')
plt.show()

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='median_income',
    y='median_house_value',
    size='housing_median_age',
    palette='viridis',
    sizes=(20, 200)
)

plt.title('Median House Value vs Median Income with Housing Median Age')
plt.xlabel('Median Income')
plt.ylabel('Median House Value')
plt.show()

plt.figure(figsize=(10, 6))

sns.regplot(
    data=df,
    x='median_income',
    y='median_house_value',
    scatter_kws={'color': 'blue'},
    line_kws={'color': 'red'}
)

plt.title('Median House Value vs Median Income with Regression Line')
plt.xlabel('Median Income')
plt.ylabel('Median House Value')
plt.show()

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='median_income',
    y='median_house_value',
    color='blue'
)

plt.title('Median House Value vs Median Income')
plt.xlabel('Median Income')
plt.ylabel('Median House Value')

plt.savefig('median_house_value_vs_median_income.png')

print("\nScatter plot saved as 'median_house_value_vs_median_income.png'")
print("\nProgram Executed Successfully!")
```
