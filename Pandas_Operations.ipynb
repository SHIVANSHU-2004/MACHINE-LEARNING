```python
# Step 1: Import libraries
import pandas as pd
import numpy as np

# Step 2: Create a sample dataset
data = {
    "Name": ["Alice", "Bob", "Carol", "Dave", "Eve", "Bob"],
    "Age": [20, 21, np.nan, 19, 22, 21],
    "Marks": [85, np.nan, 88, 90, 95, np.nan],
    "Enrollment Date": ["2023-01-10", "2023-01-12", "2023/01/15", "15-01-2023", "2023-01-18", "2023-01-12"]
}

df = pd.DataFrame(data)
print("Original Dataset:\n", df)

# Step 3: Data Cleaning Operations

# 3a. Remove duplicate rows
df.drop_duplicates(inplace=True)

# 3b. Rename columns for consistency
df.rename(columns={"Marks": "Exam_Marks", "Enrollment Date": "Enroll_Date"}, inplace=True)

# 3c. Handle missing values
# Fill missing Age with mean
df['Age'].fillna(df['Age'].mean(), inplace=True)
# Fill missing Exam_Marks with median
df['Exam_Marks'].fillna(df['Exam_Marks'].median(), inplace=True)

# 3d. Correct data types
df['Age'] = df['Age'].astype(int)
df['Enroll_Date'] = pd.to_datetime(df['Enroll_Date'], dayfirst=True, errors='coerce')  # standardize date format

# 3e. Handle outliers (optional)
df = df[df['Exam_Marks'] <= 100]  # remove unrealistic marks > 100

# Step 4: Data Preprocessing
# Create a new column 'Grade' based on Exam_Marks
def assign_grade(marks):
    if marks >= 90:
        return 'A'
    elif marks >= 80:
        return 'B'
    elif marks >= 70:
        return 'C'
    else:
        return 'D'

df['Grade'] = df['Exam_Marks'].apply(assign_grade)

# Step 5: Display cleaned and preprocessed data
print("\nCleaned and Preprocessed Dataset:\n", df)
```

```python
import pandas as pd

# Step 1: Create a small dataset
data = {
    "Name": ["Alice", "Bob", "Carol", "Dave", "Eve"],
    "Age": [20, 21, None, 19, 22],
    "Marks": [85, None, 88, 90, 95]
}

df = pd.DataFrame(data)
print("Original Dataset:\n", df)

# Step 2: Handle missing values
df['Age'].fillna(df['Age'].mean(), inplace=True)       # Fill missing age with mean
df['Marks'].fillna(df['Marks'].median(), inplace=True) # Fill missing marks with median

# Step 3: Data Preprocessing - Create a grade column
def grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    else:
        return "C"

df['Grade'] = df['Marks'].apply(grade)

# Step 4: Display cleaned and preprocessed dataset
print("\nCleaned & Preprocessed Dataset:\n", df)
```
